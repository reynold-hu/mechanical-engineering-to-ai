## 1. KNN (k = 1)

```python
def knn(X_train,Y_train,X_test,Y_test):
    knn = KNeighborsClassifier(n_neighbors=1)
    knn.fit(X_train,Y_train)
    pred = knn.predict(X_test)
    test=np.array(Y_test)
    y=[]
    for i in range(len(test)):
        y.append(test[i][0])
    error = np.mean(pred != y)*100
    return error
```

```python
error1 = knn(X_train1,Y_train1,X_test1,Y_test1)
error2 = knn(X_train2,Y_train2,X_test2,Y_test2)
error3 = knn(X_train3,Y_train3,X_test3,Y_test3)
error4 = knn(X_train4,Y_train4,X_test4,Y_test4)
error5 = knn(X_train5,Y_train5,X_test5,Y_test5)
```

```python
## LBP
def get_pixel(img, center, x, y):   
    new_value = 0  
    try:
        if img[x][y] >= center:
            new_value = 1        
    except:
        pass 
    return new_value
   
# Function for calculating LBP
def lbp_calculated_pixel(img, x, y):
    center = img[x][y]
    val_ar = [] 
    # top_left
    val_ar.append(get_pixel(img, center, x-1, y-1))
    # top
    val_ar.append(get_pixel(img, center, x-1, y))
    # top_right
    val_ar.append(get_pixel(img, center, x-1, y + 1))
    # right
    val_ar.append(get_pixel(img, center, x, y + 1))
    # bottom_right
    val_ar.append(get_pixel(img, center, x + 1, y + 1))
    # bottom
    val_ar.append(get_pixel(img, center, x + 1, y))
    # bottom_left
    val_ar.append(get_pixel(img, center, x + 1, y-1))
    # left
    val_ar.append(get_pixel(img, center, x, y-1))
    power_val = [1, 2, 4, 8, 16, 32, 64, 128]
    val = 0 
    for i in range(len(val_ar)):
        val += val_ar[i] * power_val[i]   
    return val
```

```python
img_lbp = np.zeros((32, 32),np.uint8)
cols=["Index"]
cols.extend([str(x) for x in range(0,1024)])
df = pd.DataFrame(data=np.zeros((len(images),1025)),columns=cols)
for k in tqdm(range(len(images))):   
    for i in range(0, 32):
        for j in range(0, 32):
            img_lbp[i, j] = lbp_calculated_pixel(images[k], i, j)
    df.loc[i,"Index"] = k
    lbp = img_lbp.flatten()
    df.loc[k,cols[1::]] = lbp 
```

```python
df = df.drop('Index', 1)
df
```

```python
eigenvalues, eigenvectors, = np.linalg.eig(cov_matrix)
print('Eigenvectors of Cov(X):')
print(eigenvectors)
print('Eigenvalues of Cov(X):',eigenvalues)
```

## LDA

```python
def lda(X_train,Y_train,X_test,Y_test):
    lda = LinearDiscriminantAnalysis()
    lda.fit(X_train,Y_train)
    pred = lda.predict(X_test)
    test=np.array(Y_test)
    y=[]
    for i in range(len(test)):
        y.append(test[i][0])
    error = np.mean(pred != y)*100
    return error
```
